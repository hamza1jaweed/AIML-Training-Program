```python
## Strings
'''
A string is a collection of one or more characters put in a single quote,
 double-quote or triple quote. In python there is no character data type,
 a character is a string of length one. It is represented by str class.

String can have special characters. String can be indexed

'''


name = 'First Lab'
name_extended = name + 'Module 1'
last_element_string = name[-1] # -1 in python is index of the last element.
## indexing is important for preprocessing of the raw data.
```

```python
## List

'''
Lists are ordered collection of data, and are very similar to arrays,
It is very flexible as the items in a list do not need to be of the same type.
'''

name_list = ['First Lab', 3 , '1.1' , 'Lab 1'] ## notice elements are of different data type.
name_list.extend(['Module 1']) ## adding elements to list (Read about append method as well).
element_2 = name_list[1] ## Just like other languages, the index starts from 0.
two_dimesional_list = [[1,2],[3,4]] ## practice with multi-dimensional lists and arrays
## you would soon be required to handle 4 dimensional data :p :)
name_list[2] = '1.111' ##list elements can be changed
```

```python
print(name_list)
print(element_2)
print(two_dimesional_list)
## list can have list, dictionary, string etc.
```

```python
## Tuples

name_tuple = ('First Lab', 1, (2,3),[1,1,'list having string']) ## A tuple can have a tuple.

print(name_tuple[2])
print("first indexing the last element of the tuple, which is a list and \n then last element of the list (a string) and then second last element of the string:")
print(name_tuple[-1][-1][-2])
```

```python
## Tuples are immutable, so their elements cannot be changed.
## They are useful when passing parameters that should not be modified.

try:
    name_tuple[1] = 2
except TypeError:
    print("Tuples are immutable and cannot be changed.")
```

```python
## Sets
'''a Set is an unordered collection of data types that is iterable, mutable and has no duplicate elements.
The order of elements in a set is undefined though it may consist of various elements.
The major advantage of using a set, as opposed to a list,
 is that it has a highly optimized method for checking whether a specific element is contained in the set.
'''
set_unique = set([1,1,2,3,5,6,'Lab1'])
print(set_unique) ##notice it is unordered
last_el = set_unique.pop()
set_unique.add((1,2))


print(last_el)
print(set_unique)




```

```python
## Sets are unordered and cannot be accessed using an index.

try:
    print(set_unique[1])
except TypeError:
    print("Sets are not indexable. You cannot access set elements using an index.")
```

```python
## Dictionary
'''
Dictionary in Python is an unordered collection of data values, used to store data values like a map,
 which, unlike other data types which hold only a single value as an element.
'''

dic = {'1': 'A','2':'B', 'C':3 } ##Observe how key and values can be anything
dic['4'] ='New'
print(dic)
```

```python
def check_even_list(num_list):

    even_numbers = []

    # Go through each number
    for number in num_list:
        # Once we get a "hit" on an even number, we append the even number
        if number % 2 == 0:
            even_numbers.append(number)
        # Don't do anything if its not even
        else:
            pass
    # Notice the indentation! This ensures we run through the entire for loop
    return even_numbers
```

##### Numpy - One of the most used libraries - supports for large, multi-dimensional arrays and matrices, along with a large collection of high-level mathematical functions to operate on these arrays.

```python
import numpy as np
```

```python

a = np.array([1,1,2,3,4,5,5,6,1]) ## np.array converts given list to array

b = a>1 ## important comparison operation, where frequently used in manipulation and image processing.

print(b)
print(a[b]) ## [printing only those values in a which are greater than 1]
```

```python
a_range = np.arange(10,19).reshape(3,3) ## create a 3x3 array with values in range 10-19
a_range
```

```python
## Indexing in arrays works same as that of list

a_range[0] # printing all the columns of first row
```

```python
a_range[:,2] #printing all the rows of second column
```

```python
iden = np.eye(3) #idnetity matrix of given size
iden
```

```python
## adding two matrices
summed = a_range + iden
summed
```

```python
### arrays support normal matrix multiplication that you are used to, point-wise multiplication
### and dot product as well.

mul = a_range@iden ## normal multiplication
mul
```

```python
## point wise multiplication
p_mul = a_range * iden
p_mul
```

```python
## Transpose of a matrix.

mtx_t = mul.T
mtx_t
```

```python
### Here we are changing the values of last row of the transposed matrix.
### basically point wise multiplying the values of last row with 1,2 and 3

mtx_t[2] = mtx_t[2]*[1,2,3] ## indexing, point wise multiplication and mutation of values
mtx_t
```

```python
## Just like the greater than 1 (a>1) example we saw earlier.
## here we are checking if the elements are divisible by 2 (%), and if they are, then replace by 0.

mtx_t[(mtx_t % 2 == 0)] = 0 ## convert even elements of the matrix to zero.
mtx_t
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert image to grayscale using a common luminance formula
grayscale_img = np.dot(img[...,:3], [0.2989, 0.5870, 0.1140]).astype(np.uint8)

plt.figure(figsize=(8, 8))
plt.imshow(grayscale_img, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Flip the image horizontally
flipped_img = np.fliplr(img)

plt.figure(figsize=(8, 8))
plt.imshow(flipped_img)
plt.title('Horizontally Flipped Image')
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Rotate the image 90 degrees clockwise
rotated_img = np.rot90(img)

plt.figure(figsize=(8, 8))
plt.imshow(rotated_img)
plt.title('Rotated Image (90 degrees clockwise)')
plt.axis('off')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Crop a section of the image (e.g., center 50%)
# Calculate dimensions
height, width, _ = img.shape
start_row, end_row = int(height * 0.25), int(height * 0.75)
start_col, end_col = int(width * 0.25), int(width * 0.75)

# Perform crop
cropped_img = img[start_row:end_row, start_col:end_col]

plt.figure(figsize=(8, 8))
plt.imshow(cropped_img)
plt.title('Cropped Image (Center 50%)')
plt.axis('off')
plt.show()
```
